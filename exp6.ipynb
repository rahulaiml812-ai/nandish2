```python
import numpy as np

demo1 = np.array([1, 2, 3, 4, 5])
demo2 = np.array([4, 3, 2, 1, 5])

res_map = np.array(list(map(lambda x, y: x + y, demo1, demo2)))

print("result of map using numpy", res_map)
```

```python
import numpy as np
from functools import reduce

demo1 = np.array([1, 2, 3, 4, 5])
demo2 = np.array([4, 3, 2, 1, 5])

res_reduce = reduce(lambda x, y: x + y, demo2)

print("result of reduce using numpy", res_reduce)
```

```python
import pandas as pd

data = {'A': [420, 380, 390], 'B': [50, 40, 45]}

df = pd.DataFrame(data)

df['A-mapped'] = df['A'].map(lambda x: x + 2)

print("mapped values", df['A-mapped'])
```

```python
import pandas as pd

data = {'A': [420, 380, 390], 'B': [50, 40, 45]}

df = pd.DataFrame(data)

print(df)

df['Sum'] = df.apply(lambda row: row['A'] + row['B'], axis=1)

print("Sum of A & B using apply:", df['Sum'])
```

```python
import pandas as pd
from functools import reduce

data = {'a': [20, 30, 80, 90], 'b': [50, 40, 45]}
df = pd.DataFrame(data)

print(df)

res_reduce = reduce(lambda x, y: x + y, df['b'])
print(res_reduce)
```
