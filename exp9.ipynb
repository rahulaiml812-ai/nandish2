```python
import pandas as pd 
```

```python
df=pd.read_csv('titanic.csv',usecols=['Age','Fare','Survived','Cabin'])
df
```

```python
df.isnull().sum()
```

```python
df['Survived'].fillna(1,inplace=True)
df.isnull().sum()
```

```python
print(df.columns)
```

```python
df['Fare'] = df['Fare'].ffill()
df.isnull().sum()
```

```python
df["Cabin"] = df["Cabin"].bfill()

```

```python
print(df.isnull().sum())
```

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
df.isnull().sum()
```
